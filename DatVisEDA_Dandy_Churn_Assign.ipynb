### Pembersihan Data: Mengatasi Missing Values dan Inkonsistensi

Langkah-langkah berikut akan memastikan data siap untuk visualisasi:

```python
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')
print(f"Jumlah NaN di 'Total Charges' setelah konversi: {df['Total Charges'].isnull().sum()}")

median_total_charges = df['Total Charges'].median()

df['Total Charges'].fillna(median_total_charges, inplace=True)
print("Nilai NaN di 'Total Charges' telah diisi dengan median.")

df['Churn Reason'].fillna('Not Churned', inplace=True)
print("Nilai NaN di 'Churn Reason' telah diisi dengan 'Not Churned'.")
```

```python
columns_to_drop = []


if df['Count'].nunique() == 1:
    columns_to_drop.append('Count')
    print("Kolom 'Count' adalah konstan, akan dihapus.")


if df['Country'].nunique() == 1:
    columns_to_drop.append('Country')
    print("Kolom 'Country' adalah konstan, akan dihapus.")


if df['State'].nunique() == 1:
    columns_to_drop.append('State')
    print("Kolom 'State' adalah konstan, akan dihapus.")


columns_to_drop.append('CustomerID')
print("Kolom 'CustomerID' akan dihapus.")

columns_to_drop.append('Lat Long')
print("Kolom 'Lat Long' akan dihapus.")


df.drop(columns=columns_to_drop, inplace=True)
print(f"Kolom {columns_to_drop} telah dihapus.")

print("\nDataFrame Information setelah pembersihan:")
df.info()
```

### Identifikasi Korelasi Antar Variabel Numerik


```python
# Memilih variabel numerik yang relevan untuk analisis korelasi
numerical_cols = [
    'Tenure Months',
    'Monthly Charges',
    'Total Charges',
    'Churn Value',
    'Churn Score',
    'CLTV'
]

# Menghitung matriks korelasi
correlation_matrix = df[numerical_cols].corr()

# Visualisasi dengan Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Heatmap Matriks Korelasi Variabel Numerik')
plt.show()
```

Dari *heatmap* di atas, dapat diamati bahwa ada beberapa korelasi antar variabel:

*   **Tenure Months** memiliki korelasi positif yang kuat dengan **Total Charges** (ini wajar, semakin lama berlangganan, total biaya semakin tinggi).
*   **Tenure Months** dan **Monthly Charges** memiliki korelasi negatif dengan **Churn Value** dan **Churn Score**, menunjukkan bahwa pelanggan yang lebih lama berlangganan dan/atau dengan biaya bulanan yang lebih rendah cenderung tidak *churn*.
*   **Monthly Charges** juga memiliki korelasi positif yang cukup kuat dengan **Total Charges**.
*   **Churn Value** dan **Churn Score** memiliki korelasi positif yang sangat kuat satu sama lain, yang wajar karena keduanya merepresentasikan *churn*.
*   **CLTV** memiliki korelasi negatif yang sedang dengan **Churn Value** dan **Churn Score**, yang berarti pelanggan dengan CLTV yang lebih tinggi cenderung tidak *churn*.

Letsgo lanjut eksplorasi beberapa pasangan variabel dengan korelasi kuat menggunakan *scatter plot*.

#Bukti bahwa biaya bulanan yang lebih tinggi meningkatkan churn

Tinjauan pustaka sistematis (systematic literature review) terhadap 37 artikel churn di sektor telekomunikasi (1999–2022) menemukan bahwa “monthly bill amounts” (biaya bulanan) merupakan salah satu prediktor perilaku churn paling populer, dan semua studi yang menguji variabel ini melaporkan efek positif signifikan: semakin tinggi pengeluaran bulanan, semakin tinggi probabilitas churn.

Ribeiro, H., Barbosa, B., Moreira, A.C. et al. Determinants of churn in telecommunication services: a systematic literature review. Manag Rev Q 74, 1327–1364 (2024). https://doi.org/10.1007/s11301-023-00335-7

Penelitian empiris pada prediksi churn telekomunikasi melaporkan korelasi positif kecil namun konsisten antara Monthly Charges dan Churn Score (sekitar 0,13), yang ditafsirkan sebagai indikasi bahwa kenaikan biaya bulanan meningkatkan kecenderungan churn.

Dou, Y. (2024). Customer churn prediction: An empirical research of telecommunications service provider in the United States. In Proceedings of the 1st International Conference on Data Analysis and Machine Learning (DAML 2023) (pp. 509–514). SCITEPRESS. https://doi.org/10.5220/0012802000003885

Analisis feature importance pada model churn telco menunjukkan bahwa fitur Monthly Charge dan Total Charges termasuk prediktor signifikan; pelanggan dengan biaya bulanan lebih tinggi cenderung lebih banyak yang churn, terutama pada masa berlangganan pendek.

Hermawan, A., Jayanti, N. R., Tabaruk, Z., Triadi, F. L. Y., Saputra, A., & Syachrudin, M. R. H. (2024). Membangun model prediksi churn pelanggan yang akurat: Studi kasus tentang TELCO Company. Merkurius: Jurnal Riset Sistem Informasi dan Teknik Informatika, 2(6), 67–81. https://doi.org/10.61132/merkurius.v2i6.398

## Pertanyaan 1: Bagaimana Lama Berlangganan (Tenure Months) Memengaruhi Tingkat Churn Pelanggan?

Untuk menjawab pertanyaan ini, kita akan mengelompokkan pelanggan berdasarkan 'Tenure Months' menjadi beberapa kategori (misalnya, "Baru", "Menengah", "Lama") dan menghitung tingkat churn untuk setiap kategori.

```python
# Membuat kategori 'Tenure Months'
def create_tenure_group(tenure):
    if tenure <= 12: # 1 tahun
        return '0-12 Bulan (Baru)'
    elif tenure <= 24: # 1-2 tahun
        return '13-24 Bulan (Menengah)'
    elif tenure <= 48: # 2-4 tahun
        return '25-48 Bulan (Setia)'
    else: # > 4 tahun
        return '49+ Bulan (Sangat Setia)'

df['Tenure Group'] = df['Tenure Months'].apply(create_tenure_group)

# Menghitung churn rate per Tenure Group
churn_by_tenure_group = df.groupby('Tenure Group')['Churn Value'].mean().reset_index()
churn_by_tenure_group['Churn Rate'] = churn_by_tenure_group['Churn Value'] * 100

# Mengurutkan berdasarkan Tenure Group untuk visualisasi yang lebih baik
tenure_order = ['0-12 Bulan (Baru)', '13-24 Bulan (Menengah)', '25-48 Bulan (Setia)', '49+ Bulan (Sangat Setia)']
churn_by_tenure_group['Tenure Group'] = pd.Categorical(churn_by_tenure_group['Tenure Group'], categories=tenure_order, ordered=True)
churn_by_tenure_group = churn_by_tenure_group.sort_values('Tenure Group')

display(churn_by_tenure_group)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Tenure Group', y='Churn Rate', data=churn_by_tenure_group, palette='coolwarm')
plt.title('Tingkat Churn Berdasarkan Kelompok Lama Berlangganan (Tenure Months)')
plt.xlabel('Kelompok Lama Berlangganan')
plt.ylabel('Tingkat Churn (%)')
plt.ylim(0, churn_by_tenure_group['Churn Rate'].max() * 1.1)

# Menambahkan nilai churn rate di atas bar
for index, row in churn_by_tenure_group.iterrows():
    plt.text(index, row['Churn Rate'] + 0.5, f"{row['Churn Rate']:.2f}%", color='black', ha="center")

plt.tight_layout()
plt.show()
```

### Insight dari Pertanyaan Analisis 1:
Dari visualisasi tingkat churn berdasarkan kelompok lama berlangganan, terlihat jelas bahwa pelanggan baru (0-12 Bulan) memiliki tingkat churn yang paling tinggi. Tingkat churn cenderung menurun secara signifikan seiring dengan bertambahnya lama berlangganan pelanggan. Pelanggan yang telah berlangganan 49 bulan atau lebih (Sangat Setia) menunjukkan tingkat churn yang sangat rendah.

Hal ini mengindikasikan bahwa periode awal berlangganan adalah fase kritis di mana pelanggan paling rentan untuk churn. Strategi retensi harus lebih difokuskan pada pelanggan baru untuk meningkatkan loyalitas mereka, misalnya dengan program onboarding yang kuat, dukungan pelanggan yang responsif, atau insentif khusus di bulan-bulan pertama.

Strategi retensi dapat mencakup penawaran insentif untuk mendorong pelanggan beralih ke kontrak jangka panjang atau meningkatkan kepuasan pelanggan secara signifikan pada pelanggan 'Month-to-month'.

## Pertanyaan 4: Bagaimana Biaya Bulanan (Monthly Charges) Memengaruhi Tingkat Churn Pelanggan?

Untuk menjawab pertanyaan ini, kita akan mengelompokkan 'Monthly Charges' ke dalam beberapa kategori (misalnya, 'Rendah', 'Menengah', 'Tinggi') dan menghitung tingkat churn untuk setiap kategori tersebut.

```python
# Membuat kategori untuk 'Monthly Charges'
def create_monthly_charge_group(charge):
    if charge <= 30:
        return 'Rendah (<30$)'
    elif charge <= 70:
        return 'Menengah (30$-70$)'
    else:
        return 'Tinggi (>70$)'

df['Monthly Charge Category'] = df['Monthly Charges'].apply(create_monthly_charge_group)

# Menghitung churn rate per kategori Monthly Charges
churn_by_monthly_charge = df.groupby('Monthly Charge Category')['Churn Value'].mean().reset_index()
churn_by_monthly_charge['Churn Rate'] = churn_by_monthly_charge['Churn Value'] * 100

# Mengurutkan berdasarkan kategori untuk visualisasi yang lebih baik
mc_order = ['Rendah (<30$)', 'Menengah (30$-70$)', 'Tinggi (>70$)']
churn_by_monthly_charge['Monthly Charge Category'] = pd.Categorical(churn_by_monthly_charge['Monthly Charge Category'], categories=mc_order, ordered=True)
churn_by_monthly_charge = churn_by_monthly_charge.sort_values('Monthly Charge Category')

display(churn_by_monthly_charge)
```

```python
plt.figure(figsize=(9, 6))
sns.barplot(x='Monthly Charge Category', y='Churn Rate', data=churn_by_monthly_charge, palette='autumn')
plt.title('Tingkat Churn Berdasarkan Kategori Biaya Bulanan')
plt.xlabel('Kategori Biaya Bulanan')
plt.ylabel('Tingkat Churn (%)')
plt.ylim(0, churn_by_monthly_charge['Churn Rate'].max() * 1.1)

# Menambahkan nilai churn rate di atas bar
for index, row in churn_by_monthly_charge.iterrows():
    plt.text(index, row['Churn Rate'] + 0.5, f"{row['Churn Rate']:.2f}%", color='black', ha="center")

plt.tight_layout()
plt.show()
```

### Insight dari Pertanyaan Analisis 4:
Dari visualisasi tingkat churn berdasarkan kategori biaya bulanan, terlihat bahwa pelanggan dengan biaya bulanan 'Tinggi ($> \!70$)' memiliki tingkat churn yang paling tinggi secara signifikan. Tingkat churn menurun seiring dengan berkurangnya biaya bulanan, dengan pelanggan di kategori 'Rendah ($< \! 30$)' menunjukkan tingkat churn terendah.

Insight ini menunjukkan bahwa harga adalah faktor penting yang memengaruhi keputusan pelanggan untuk churn. Pelanggan yang membayar biaya bulanan lebih tinggi cenderung lebih kritis terhadap nilai yang mereka terima dan lebih rentan untuk mencari alternatif jika merasa tidak puas. Strategi retensi harus mempertimbangkan penawaran paket yang kompetitif untuk pelanggan berbiaya tinggi atau memberikan nilai tambah yang jelas untuk membenarkan biaya tersebut.

## Pertanyaan 5: Bagaimana Metode Pembayaran Memengaruhi Tingkat Churn Pelanggan?

Untuk menjawab pertanyaan ini, kita akan menghitung churn rate untuk setiap metode pembayaran yang digunakan oleh pelanggan.

```python
# Menghitung churn rate per metode pembayaran
churn_by_payment_method = df.groupby('Payment Method')['Churn Value'].mean().reset_index()
churn_by_payment_method['Churn Rate'] = churn_by_payment_method['Churn Value'] * 100

# Mengurutkan berdasarkan churn rate untuk visualisasi yang lebih baik
churn_by_payment_method = churn_by_payment_method.sort_values('Churn Rate', ascending=False)

display(churn_by_payment_method)
```

### Insight dari Pertanyaan Analisis 5:
Dari visualisasi tingkat churn berdasarkan metode pembayaran, terlihat bahwa pelanggan yang menggunakan 'Electronic check' memiliki tingkat churn yang paling tinggi secara signifikan. Sebaliknya, metode pembayaran otomatis seperti 'Bank transfer (automatic)' dan 'Credit card (automatic)' memiliki tingkat churn yang jauh lebih rendah. 'Mailed check' berada di antara keduanya.

Insight ini menunjukkan bahwa metode pembayaran dapat menjadi indikator risiko churn. Pelanggan yang memilih 'Electronic check' mungkin cenderung lebih aktif dalam mengelola keuangan mereka dan lebih mudah beralih ke penyedia layanan lain. Metode pembayaran otomatis sering kali menunjukkan tingkat komitmen yang lebih tinggi dari pelanggan dan mengurangi 'gesekan' pembayaran bulanan. Strategi retensi dapat mencakup mendorong pelanggan untuk beralih ke metode pembayaran otomatis melalui insentif, yang tidak hanya mengurangi churn tetapi juga mengotomatiskan proses penagihan.
